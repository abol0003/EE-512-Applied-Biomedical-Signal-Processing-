<h1>EX01: Influence of the parametrization of the Welch spectral</h1>

</center>


<br>
    The objective of this exercise is to study the influence of the<br>
    parameterization of the Welch spectral estimator in order to highlight a 50<br>
    Hz perturbation in an ECG signal.<br>


```python
import numpy as np
import pylab as py
py.ion()
py.close('all')
import scipy.signal as sp
```

```python
x = np.genfromtxt('ecg.dat')
fs = 500
```


<br>
    Objective: Compare spectral estimation for different window lengths using<br>
    welch estimation.<br>
    Plot the log spectrum of the signal using windows of 100, 500, 2000.<br>

Q: Comment the results.<br>

<span style=color:blue> As we can see the lower the window length is the lower the resolution isand then the less bins we have meaning in a higher average of the signal. Indeed, fs=500 for length =100 leads to a resolution of $$ f_r = \frac{2*f_s}{N} = \frac{1000}{100} = 10 \ \text{Hz} $$ So between 0 and 250Hz we  only have $$\frac{250-0}{10}=25 \text{bins}$$


Q: Which windows length is the most suitable for the observation of 50 Hz?<br>
<span style=color:blue>The most suitable window length is the 500 sample length.


Q: Why?<br>
 <span style=color:blue>
As we can see from the plots is that the window of length 100 is averaging to much the signal to see the 50Hz peaks due to the bad resolution. The 2000 sample window almost not average anithing making a very noisy signal where we cannot extract suitable information at 50Hz. The 500 sample window then make the good trade off between resolution and noise reducing by averaging and enable to see clearly the peak a 50Hz


```python
f,X_100 = sp.welch(x, nperseg=100, nfft=4096, fs=fs)
f,X_500 = sp.welch(x, nperseg=500, nfft=4096, fs=fs)
f,X_2000 = sp.welch(x, nperseg=2000, nfft=4096, fs=fs)
```

```python
py.figure(1, figsize=[5,8])
py.clf()
py.subplot(3,1,1)
py.plot(f, 10*np.log10(X_100), 'k')
py.grid()
py.xlabel('frequency (Hz)')
py.ylabel('power (dB)')
py.title('PSD with lengh = 100')
```

```python
py.subplot(3,1,2)
py.plot(f, 10*np.log10(X_500), 'k')
py.grid()
py.xlabel('frequency (Hz)')
py.ylabel('power (dB)')
py.title('PSD with lengh = 500')
```

```python
py.subplot(3,1,3)
py.plot(f, 10*np.log10(X_2000), 'k')
py.grid()
py.xlabel('frequency (Hz)')
py.ylabel('power (dB)')
py.title('PSD with lengh = 2000')
```
